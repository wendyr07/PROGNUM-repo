```python
#7.1 SESSION QUIZ
import numpy as np
from numpy import linalg as la

A = np.array([1,2,3, 0,-4,1, 0,3,-1]).reshape(3,3)
print(f"Matrix A: \n{A}")


print(f"The determinant of A is:{la.det(A)}")

b = np.array([2,2,1])
x,y,z = la.solve(A, b)
print(f" A = \n{A}")
print(f" b = {b}" )
print(f"The solution (x,y,z) for Ax = b: ({x},{y},{z}).")


print("A is orthogonal if A@A.T = I")
Atrans = A.T
print(f"The transposed matrix of A is: \n{Atrans}")
print(f"The identity matrix: {np.identity(3)}")
print(f"A@A.T == I: \n{A@A.T == np.identity(3)}")
print(f"The dot product of A and A transposed is not the same as the identity matrix, so no, A is not orthogonal.")

Lambda, v = la.eig(A)
print(f"The Eigenvalues are: {Lambda}.")
```

```python
#7.2 OPTIONAL
from math import radians, cos, sin
def Ex(angle):
    a = cos(radians(angle))
    b = sin(radians(angle))
    Ex = np.array([1,0,0, 0,a,b, 0,-b,a]).reshape(3,3)
    return Ex

def Ey(angle):
    a = cos(radians(angle))
    b = sin(radians(angle))
    Ey = np.array([a,0,-b, 0,1,0, b,0,a]).reshape(3,3)
    return Ey
def Ez(angle):
    a = cos(radians(angle))
    b = sin(radians(angle))
    Ez = np.array([a,b,0, -b,a,0, 0,0,1]).reshape(3,3)
    return Ez


print(f"\nEx(33)=\n{Ex(33)}")
print(f"\nEy(33)=\n{Ey(33)}")
print(f"\nEz(33)=\n{Ez(33)}")
```

```python
#7.3 SESSION QUIZ

E = Ez(180-123)@Ey(90-27.4)@Ez(192.25)
print(E)

print("For E to be a rotational matrix we need E.T = inv.(E) and det(E) = 1")
print(f"\nThe transposed matrix is: \n{E.T}")
print(f"The inverse of the matrix is: \n{la.inv(E)}")
print(f"E.T == E^-1: {np.allclose(E.T, la.inv(E))}")

print(f"\nThe determinant of E is: {la.det(E)}.")
print(f"Det(E) == 1: {la.det(E) == 1}.")

print(f"\nBoth the conditions are met so E is proven to be a rotational matrix.")

#Eigenvalues
Lambda, v = la.eig(E)
print(f"\nThe eigenvalues: {Lambda}.")
print(f"So there is an eigenvector that is close to one: {Lambda[2]}.")

dec = radians(27.4)
RA = radians(192.25)
r = np.array([cos(dec)*cos(RA), cos(dec)*sin(RA), sin(dec)])
print(f"\nFrom the given equation for r: r = {r}, which are inline with the given vectors.")
print(f"The norm of these vectors is: {la.norm(r)}.")

v2 = E@r
print(f"\nThe vector v2 is not exactly (0,0,1) but is very close to is with its values being: {v2}.")
```

```python
#7.4
import numpy as np
#Data:
x = np.array([1.2, 2.0, 3.1, 4.4, 4.9, 5.8, 6.9, 8.0, 9.1, 9.9])
y = np.array([0.3, 0.5, 0.7, 0.8, 1.3, 1.6, 2.0, 2.3, 2.9, 3.0])

X = float(input("Which value for x?"))

def li(x,y,X):
    for i in range(len(x)-1): #-1 since later on we use the i+1 and otherwise those we be out of bounds.
        if x[i] <= X <= x[i+1]: #Finds the two neighbours of X.
            y = y[i] + ((y[i+1] - y[i])*(X-x[i]))/ (x[i+1]-x[i]) #Linear polation function.
    return y #Returns y.

#Printing.
print(f"\nThe y for X = 2.5 is: {li(x,y,2.50):.2f}, using the formula.")
print(f"The y for X = {X} is: {li(x,y,X):.2f}, using the formula.")

#Printing others.
print(f"\nThe y for X 2.5 according to numpy is: {np.interp(2.5, x, y):.2f}.")
print(f"The y for X = {X} according to numpy is: {np.interp(X,x,y):.2f}.")
```

```python
#7.5
from numpy import linalg as la
from matplotlib import pyplot as plt

x = np.array([1.2, 2.0, 3.1, 4.4, 4.9, 5.8, 6.9, 8.0, 9.1, 9.9])
y = np.array([0.3, 0.5, 0.7, 0.8, 1.3, 1.6, 2.0, 2.3, 2.9, 3.0])
n = len(x) #Makes it less typing and thus less sources of error further in this program.
A = np.zeros([n,n]) #Makes a matrix with all zeros, which will be replaced later on.

for i in range(n): #Goes through all of the rows.
    for j in range(n): #Goes through all of the columns.
        A[i,j] = x[i]**(n-j-1) #Calculates the matrix entry.

solutions = la.solve(A,y) #Solves for the coefficients.
print(solutions)

#Polynomial
x_pol = np.linspace(1,10,10*n) #Makes the x for the polynomial, so it looks smoother than just the x points from the data.
y_pol = np.polyval(solutions, x_pol) #Evaluates the polynomial at said x values.

#Plotting the graph.
fig = plt.figure(figsize = (8,6))
frame = fig.add_subplot(1,1,1)
plt.plot(x_pol, y_pol, label = "Polynomial", color = "cornflowerblue")
plt.scatter(x,y, label="data points", color = "mediumorchid")
frame.legend()
plt.show()
```
